```python
class Building:
  def __init__ (self, height, floor_num, city = "Kyiv"):
    self.height = height
    self.floor_num = floor_num
    self.city = city
  def is_tall (self):
    if self.height > 50:
      return True
    else:
      return False
  def floor_height (self):
    return self.height / self.floor_num
```

Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math
class Cylinder(Shape3D):
  def __init__(self, height, radius):
    self.height = height
    self.radius = radius
  def get_volume(self):
    return math.pi * self.radius**2 * self.height
class Cube(Shape3D):
  def __init__(self, side):
    self.side = side
  def get_volume(self):
    return self.side**3
cubic1 = Cube(3)
cylinder1 = Cylinder (5, 3)
print (cubic1.get_volume() + cylinder1.get_volume())
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
  def __init__(self, wallet_id):
    self.__balance = 0
    self.__id = wallet_id
  @property
  def balance(self):
    return self.__balance
  @property
  def wallet_id(self):
    return self.__id
  @balance.setter
  def balance(self, value):
    if value < 0:
        return ("баланс не може бути від'ємним")
    self.__balance = value
  def deposit(self, money):
    if money > 0:
      self.balance = self.balance + money
    else:
      print ("поповнення не може бути на від'ємну суму")
  def withdrawal(self, money):
    if money <= 0:
      print ("зняття не може бути на від'ємну суму")
    elif money > self.balance:
      print ("не можна зняти більше грошей ніж доступно")
    else:
      self.balance = self.balance - money
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
  def __init__(self, name, price):
    self.name = name
    self.price = price
class ShoppingCart:
  def __init__(self):
    self.items = []
  def add_item (self, item):
    self.items.append(item)
  def delete_item (self, item):
    if item in self.items:
      self.items.remove(item)
  def show_item_list(self):
    for item in self.items:
      print("item name:", item.name)
      print("item price: ", item.price)
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class Week:
  def __init__(self):
    self.days = ["понеділок", "вівторок", "середа", "четвер", "п'ятниця", "субота", "неділя"]
    self.index = 0
  def __iter__(self):
    return self
  def __next__(self):
    if self.index < len(self.days):
      day = self.days[self.index]
      self.index += 1
      return day
    else:
      print ("більше нема днів")
week = Week()
it = iter(week)
print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```
